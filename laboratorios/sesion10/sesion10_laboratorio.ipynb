````python
%%writefile artesano_sesion10_c.c
/* artesano_sesion10_c.c - Fase 1: El Artesano (sin IA)
 * Mini-evaluador de expresiones (Hito 1).
 *
 * Implementa lo que declara api_sesion10.h.
 * El harness compila este archivo junto con el suyo: no escribas main().
 */

#include "api_sesion10.h"
#include <stdio.h>
#include <ctype.h>
#include <limits.h>

typedef struct {
    const char *texto;
    int error;
} Parser;

void saltar(Parser *p);
long primaria(Parser *p);
long termino(Parser *p);
long expr(Parser *p);

/* Saltar espacios */
void saltar(Parser *p) {
    while (isspace((unsigned char)*p->texto)) {
        p->texto++;
    }
}

/* Leer numeros y parentesis */
long primaria(Parser *p) {
    long num = 0;

    saltar(p);

    if (*p->texto == '(') {
        p->texto++;

        num = expr(p);
        saltar(p);

        if (*p->texto == ')') {
            p->texto++;
        } else {
            p->error = 2;
        }

        return num;
    }

    if (!isdigit((unsigned char)*p->texto)) {
        p->error = 2;
        return 0;
    }

    while (isdigit((unsigned char)*p->texto)) {
        int digito = *p->texto - '0';

        if (num > (LONG_MAX - digito) / 10) {
            p->error = 2;
            return 0;
        }

        num = num * 10 + digito;
        p->texto++;
    }

    return num;
}

/* Multiplicacion y division */
long termino(Parser *p) {
    long num = primaria(p);

    saltar(p);

    while (*p->texto == '*' || *p->texto == '/') {
        char operador = *p->texto;
        long num2;

        p->texto++;
        num2 = primaria(p);

        if (p->error != 0) {
            return 0;
        }

        if (operador == '*') {
            num = num * num2;
        } else {
            if (num2 == 0) {
                p->error = 1;
                return 0;
            }

            if (num == LONG_MIN && num2 == -1) {
                p->error = 2;
                return 0;
            }

            num = num / num2;
        }

        saltar(p);
    }

    return num;
}

/* Suma y resta */
long expr(Parser *p) {
    long num = termino(p);

    saltar(p);

    while (*p->texto == '+' || *p->texto == '-') {
        char operador = *p->texto;
        long num2;

        p->texto++;
        num2 = termino(p);

        if (p->error != 0) {
            return 0;
        }

        if (operador == '+') {
            num = num + num2;
        } else {
            num = num - num2;
        }

        saltar(p);
    }

    return num;
}

/* Funcion principal del API */
long evaluar_expresion(const char *fuente, int *error) {
    Parser p;
    long resultado;

    if (error != NULL) {
        *error = 0;
    }

    if (fuente == NULL) {
        if (error != NULL) {
            *error = 2;
        }
        return 0;
    }

    p.texto = fuente;
    p.error = 0;

    resultado = expr(&p);

    saltar(&p);

    if (*p.texto != '\0' && p.error == 0) {
        p.error = 2;
    }

    if (error != NULL) {
        *error = p.error;
    }

    if (p.error != 0) {
        return 0;
    }

    return resultado;
}

````

## Fase 2 · El Científico (10 min)

Redacta el contrato **C-R-E-O** (Contexto, Rol, Especificación, Output) para pedirle
la misma función a la IA, y pega el resultado en la celda de `ia_sesion10_c.c`. La
firma debe ser idéntica a la del contrato; si la IA la cambia, corrígela antes de

### Contrato C-R-E-O

- **C (Contexto)**: Analizador sintáctico por descenso recursivo en C (`expr`, `termino`, `primaria`) para evaluar expresiones aritméticas.
- **R (Rol)**: Experto en compiladores y lenguaje C.
- **E (Especificación)**:
  - Gramática: `expr -> termino ((+|-)...)*`, `termino -> primaria ((*|/)...)*`, `primaria -> '(' expr ')' | digito+`.
  - Ignorar espacios mediante `saltar(p)` antes de consumir tokens.
  - Guardar códigos de error en `p->error`: `1` para división entre cero y `2` para errores sintácticos (retornando `0`).
  - Compilación limpia con `-Wall -Wextra -Werror` y sin función `main()`.
- **O (Output)**: Implementación lista para usar en `ia_sesion10_c.c` que sea compatible con la cabecera `api_sesion10.h`.

````python
%%writefile ia_sesion10_c.c
/* ia_sesion10_c.c - Fase 2: El Cientifico
 * Mini-evaluador de expresiones robusto ante punteros nulos.
 */

#include "api_sesion10.h"
#include <ctype.h>
#include <limits.h>
#include <stddef.h>

typedef struct {
    const char *actual;
    int error;
} Parser;

static void saltar(Parser *p);
static long expr(Parser *p);
static long termino(Parser *p);
static long primaria(Parser *p);

/* Ignorar espacios */
static void saltar(Parser *p) {
    while (*p->actual != '\0' &&
           isspace((unsigned char)*p->actual)) {
        p->actual++;
    }
}

/* Leer numeros y parentesis */
static long primaria(Parser *p) {
    long valor = 0;

    saltar(p);

    if (p->error != 0) {
        return 0;
    }

    if (*p->actual == '(') {
        p->actual++;

        valor = expr(p);

        if (p->error != 0) {
            return 0;
        }

        saltar(p);

        if (*p->actual != ')') {
            p->error = 2;
            return 0;
        }

        p->actual++;
        return valor;
    }

    if (!isdigit((unsigned char)*p->actual)) {
        p->error = 2;
        return 0;
    }

    while (isdigit((unsigned char)*p->actual)) {
        int digito = *p->actual - '0';

        if (valor > (LONG_MAX - digito) / 10) {
            p->error = 2;
            return 0;
        }

        valor = valor * 10 + digito;
        p->actual++;
    }

    return valor;
}

/* Multiplicacion y division */
static long termino(Parser *p) {
    long valor = primaria(p);

    while (p->error == 0) {
        char operador;
        long siguiente;

        saltar(p);
        operador = *p->actual;

        if (operador != '*' && operador != '/') {
            break;
        }

        p->actual++;
        siguiente = primaria(p);

        if (p->error != 0) {
            return 0;
        }

        if (operador == '*') {
            valor *= siguiente;
        } else {
            if (siguiente == 0) {
                p->error = 1;
                return 0;
            }

            if (valor == LONG_MIN && siguiente == -1) {
                p->error = 2;
                return 0;
            }

            valor /= siguiente;
        }
    }

    return valor;
}

/* Suma y resta */
static long expr(Parser *p) {
    long valor = termino(p);

    while (p->error == 0) {
        char operador;
        long siguiente;

        saltar(p);
        operador = *p->actual;

        if (operador != '+' && operador != '-') {
            break;
        }

        p->actual++;
        siguiente = termino(p);

        if (p->error != 0) {
            return 0;
        }

        if (operador == '+') {
            valor += siguiente;
        } else {
            valor -= siguiente;
        }
    }

    return valor;
}

/* Funcion principal del API */
long evaluar_expresion(const char *fuente, int *error) {
    Parser p;
    long resultado;

    if (error != NULL) {
        *error = 0;
    }

    if (fuente == NULL) {
        if (error != NULL) {
            *error = 2;
        }

        return 0;
    }

    p.actual = fuente;
    p.error = 0;

    resultado = expr(&p);

    saltar(&p);

    if (p.error == 0 && *p.actual != '\0') {
        p.error = 2;
    }

    if (error != NULL) {
        *error = p.error;
    }

    if (p.error != 0) {
        return 0;
    }

    return resultado;
}
````

## Fase 3 · El Auditor (10 min)

El harness de abajo **no aprueba nada por sí solo**: ejecuta tu función caso por caso,
la compara contra el valor esperado e imprime `OK` o `FALLA`. Léelo antes de ejecutarlo:
los casos te dicen qué se está midiendo.


````python
%%writefile harness_sesion10_c.c
/* harness_sesion10_c.c - auditoria automatica de la Sesion 10
 *
 * Compilacion:
 *   gcc -std=c11 -Wall -Wextra -Werror harness_sesion10_c.c artesano_sesion10_c.c -o test_artesano
 *
 * El harness NO aprueba nada por si solo: cada linea se calcula
 * ejecutando tu implementacion y comparandola contra el valor esperado.
 * Si algun caso falla, el programa termina con codigo de salida 1.
 */
#include <stdio.h>
#include <string.h>
#include "api_sesion10.h"

static int probadas = 0;
static int fallidas = 0;

static void verificar_int(const char *caso, long esperado, long obtenido) {
    int ok = (esperado == obtenido);
    probadas++;
    if (!ok) fallidas++;
    printf("[%2d] %-34s esperado: %-12ld obtenido: %-12ld %s\n",
           probadas, caso, esperado, obtenido, ok ? "OK" : "FALLA");
}

int main(void) {
    printf("=== AUDITORIA SESION 10 - Mini-evaluador de expresiones (Hito 1) ===\n");
    printf("Caso                               esperado:              obtenido:              resultado\n");

    { int e=0; long r=evaluar_expresion("2+3*4",&e); (void)e; verificar_int("\"2+3*4\"", (long)(14), (long)(r)); }
    { int e=0; long r=evaluar_expresion("(2+3)*4",&e); (void)e; verificar_int("\"(2+3)*4\"", (long)(20), (long)(r)); }
    { int e=0; long r=evaluar_expresion("10/0",&e); (void)r; verificar_int("\"10/0\" -> error 1", (long)(1), (long)(e)); }
    { int e=0; long r=evaluar_expresion("2+",&e); (void)r; verificar_int("\"2+\" -> error 2", (long)(2), (long)(e)); }
    { int e=0; long r=evaluar_expresion("  7 ",&e); (void)e; verificar_int("\"  7 \"", (long)(7), (long)(r)); }
    { int e=0; long r=evaluar_expresion("5-10",&e); (void)e; verificar_int("\"5-10\"", (long)(-5), (long)(r)); }

    printf("\nRESULTADO: %d/%d pruebas superadas\n",
           probadas - fallidas, probadas);
    if (fallidas != 0) {
        printf("Aun hay %d caso(s) sin resolver. Revisa tu implementacion.\n",
               fallidas);
        return 1;
    }
    printf("TODO EN VERDE. Documenta tu auditoria en auditorias/auditoria_sesion10.md" "\n");
    return 0;
}
````
